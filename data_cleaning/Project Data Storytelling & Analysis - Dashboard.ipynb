```python
import pandas as pd
import numpy as np

# Mengimport data dari csv
df=pd.read_csv(r"C:\Users\HP\Downloads\Sample_Superstore.csv", encoding='latin-1')
df
```

```python
# Data Understanding

df.info()
```

```python
df.describe()
```

```python
# Finding missing values and duplicates

missing_values = df.isnull().sum()
percentage_missing = (missing_values / len(df)) * 100
duplicated_values = df.duplicated().sum()

print("Missing Values:")
print(missing_values)

print("\nPercentage of Missing Values:")
print(percentage_missing.round(2))

print("\n" + "=" * 30)

print("Duplicated Values:", duplicated_values)
```

```python
# Checking unique in columns

for col in df.columns:
    jumlah_unik = df[col].nunique()
    print(f"Kolom {col}: {jumlah_unik} nilai unik")
```

```python
for col in df.columns:
    print(f"\n=== {col} ===")
    print(df[col].unique())
```

```python
# Data Cleaning

# Create a copy of the original DataFrame
df_clean = df.copy()
```

```python
# Mengubah order date dan ship date menjadi datetime

df_clean['Order Date'] = pd.to_datetime(df_clean['Order Date'])
df_clean['Ship Date'] = pd.to_datetime(df_clean['Ship Date'])
df_clean['Postal Code'] = df_clean['Postal Code'].astype(str)
df_clean 
```
